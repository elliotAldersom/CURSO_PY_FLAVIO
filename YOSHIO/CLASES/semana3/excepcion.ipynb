# Manejo de Errores
## Bloques 
- `try`
- `except` 
- `else`
- `finally`  

```python
texto = 'abc'
```

```python
try: 
    numero = int(texto)
except ValueError:
    print("Error: El texto ingresado no es un número válido") 

print("Fin de celda")
```

```python
try: 
    numero = float(texto)
except ValueError as e:
    print("Error: El texto ingresado no es un número válido")
    print(f"Error: {e}") 

print("Fin de celda")
```

```python
# RAISE: Disparar un error de forma intencional 
edad = -5 
try: 
    if edad <0: 
            raise ValueError("ValueError: La edad no puede ser negativa")

except KeyError as e: 
    print("error")
except ValueError as e: 
    print(f"Error: {e}")
except KeyboardInterrupt as e: 
    print("Salida forzada...")
```

```python
transacciones_raw = [
    {"id": "TX-101", "canal": "Meta Ads",   "ingresos": "1200.50", "sesiones": "60",  "descuento": "100"},
    {"id": "TX-102", "canal": "Google Ads", "ingresos": "850.00",  "sesiones": "0",   "descuento": "50"},
    {"id": "TX-103", "canal": "TikTok Ads", "ingresos": "N/A",     "sesiones": "35",  "descuento": "0"},
    {"id": "TX-104", "canal": "Meta Ads",   "ingresos": "640.00",  "sesiones": "32"},  # Falta 'descuento'
    {"id": "TX-105", "canal": "Orgánico",   "ingresos": "-200.00", "sesiones": "15",  "descuento": "0"}, # Ingreso negativo
    {"id": "TX-106", "canal": "Google Ads", "ingresos": "1500.00", "sesiones": "75",  "descuento": "150"},
]
```

```python
ventas_limpias = []
errores = []
total_por_canal = []

# Detectar errores 

for row in transacciones_raw: 
    error = None
    try: 
        id = row.get('id', 'Sin "id"')
        canal = row.get('canal', 'Sin "canal"')
        sesiones = row.get('sesiones', 'Sin "sesiones"')
        ingresos = row.get('ingresos', 0.0)
        descuento = row.get('descuento', 0.0)

        sesiones = int(sesiones)
        ingresos = float(ingresos)
        descuento = float(descuento)

        ingreso_neto = ingresos - descuento
        ingreso_sesion = 0 
        if(sesiones > 0): ingreso_sesion = ingreso_neto / sesiones 
        else: ingreso_neto = 0

        print("corroecto " + str(id))
    
    except KeyError as e: 
        error = f"Error al buscar la clave '{e}'"
    except ValueError as e: 
        error = f"Error de valor: {e}"
    except TypeError as e: 
        error = f"Error al hacer operaciones matemáticas con valores no numéricos: {e}"
    except ZeroDivisionError as e: 
        error = f"Error al dividir entre 0: {e}"
    if(error != None):
        errores.append({"motivo": error})
```
