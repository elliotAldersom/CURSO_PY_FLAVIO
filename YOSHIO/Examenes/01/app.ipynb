### 2. Limpieza de Datos Crudos

En una exportación CSV recibes los campos como texto:

```python

devoluciones_str = "420.25"
clientes_activos_str = "310"
```

**Tareas:**
- Convierte los valores a sus tipos numéricos correspondientes (`float` e `int`).
- Calcula el ingreso neto.
- Calcula el ingreso promedio por cliente activo.
- Muestra ambos resultados.

````python
ingresos_str = "12850.75"
devoluciones_str = "420.25"
clientes_activos_str = "310"

ingresos_float = float(ingresos_str)
devoluciones_float = float(devoluciones_str)
clientes_activos_int = int(clientes_activos_str) 

ingreso_neto = ingresos_float - devoluciones_float 
ingreo_promedio_cliente = ingreso_neto / clientes_activos_int 

print(f"""
El ingreso neto es: {ingreso_neto}\n 
El ingreso promedio por cliente es: {round(ingreo_promedio_cliente, 2)}
""")
````

### 3. Cálculo de KPIs de Conversión

Una página web recibió los siguientes resultados:

```python
visitas = 4850
registros = 610
compras = 145
```

**Tareas:**

1. Calcula la tasa de registro:

   `registros / visitas * 100`

2. Calcula la tasa de conversión a compra:

   `compras / registros * 100`

3. Evalúa con operadores de comparación y lógicos si se cumple la meta de:
   - Más de 10 % de registros.
   - Más de 20 % de conversión a compra.

````python
visitas = 4850
registros = 610
compras = 145

tasa_registros = (registros / visitas) * 100
tasa_compra = (compras / registros) * 100 

meta_registros = visitas * 0.1 
meta_compras = registros * 0.2 

print(f"Se cumplió la meta de registros con un % {round(tasa_registros, 2)}") if registros > meta_registros else print(f"No se completó la meta, faltó {round(meta_registros - registros)} registros")

print(f"Se cumplió la meta de compras con un % {round(tasa_compra, 2)}")  if compras > meta_compras  else print(f"No se cumplió la meta, faltó {round(meta_compras - compras)} compras")
````

# Sección 4: Bucles `while` y `for`


### 7. Ingesta y Validación por Consola (`while`)

Simula la carga manual de lotes de producción pidiendo cantidades enteras mediante `input()` hasta que el usuario ingrese `0` para finalizar.

**Tareas:**

- Si ingresa un número negativo, muestra un aviso y usa `continue` para ignorarlo sin sumarlo.
- Si ingresa `0`, finaliza el proceso.
- Al terminar, imprime:
  - La suma total de unidades válidas.
  - Cuántos lotes válidos se registraron.

````python
list_lotes = []

while(True): 
    number = int(input("Ingrese el número de lotes a registrar, o '0' para salir: "))

    if(number == 0): 
        print("saliendo del programa...")
        break

    if(number < 0): 
        print("No puede ingresar números negativos...")
        continue

    print("Número de lotes válido, registrando...") 
    list_lotes.append(number)


print(f"Número total de lotes: {sum(list_lotes)}")
print(f"Número total de lotes registrados: {len(list_lotes)}")
````

### 8. Análisis de Serie Temporal de Ventas (`for` y `range`)

Dada la lista de ventas diarias de una semana:

```python
ventas_semana = [1200, 1450, 980, 1600, 1750, 2100, 1900]
```

**Tareas:**

Recórrela usando `range(len(ventas_semana))` para imprimir:

- El número de día (del 1 al 7).
- La venta del día.
- Si estuvo por encima o por debajo de la meta diaria de 1500.


````python
ventas_semana = [1200, 1450, 980, 1600, 1750, 2100, 1900]
meta_diaria = 1500 

for i in range(len(ventas_semana)): 
    print(f"Día {i +1} | Total de venta: S/.{ventas_semana[i]} | {'meta cumplida' if ventas_semana[i] > meta_diaria else f'No se cumplió, faltó S/.{meta_diaria - ventas_semana[i]}'}")
````

### 9. Conteo de Caracteres y Códigos de Error (`for` en strings)

Dado el registro de eventos de un sistema:

```python
log_estados = "EEOOEOOOEEEOOEO"
```

Donde:
- `E` = Error.
- `O` = Operativo.

**Tareas:**

Recorre el registro con un bucle `for` para:

- Contar cuántos errores ocurrieron.
- Contar cuántas operaciones exitosas ocurrieron.
- Calcular el porcentaje de fallos del sistema.

````python
log_estados = "EEOOEOOOEEEOOEO" 
total_errores = 0
total_exitosos = 0

for i in log_estados: 
    if(i == "O"):total_exitosos += 1 
    else: total_errores += 1

porcentaje_error = (total_errores/len(log_estados)) * 100
print("==== DETALLES DEL ANÁLISIS ====")
print(f"Total de operaciones exitosas: {total_exitosos}\nTotal de operaciones fallidas: {total_errores}\nPorcentaje de error: {round(porcentaje_error,2)}%")
````

### 10. Limpieza y Ordenamiento de Lista de Precios

Tienes la siguiente lista:

```python
precios_muestra = [49.9, 120.0, 49.9, 85.5, 310.0, 15.0]
```

**Tareas:**

Aplica métodos de listas para:

a. Agregar `99.0` al final.

b. Insertar `60.0` en el índice `1`.

c. Eliminar la primera aparición de `49.9`.

d. Contar cuántas veces quedó `49.9`.

e. Ordenar la lista de mayor a menor e imprimirla.

````python
precios_muestra = [49.9, 120.0, 49.9, 85.5, 310.0, 15.0]

# agregar al final 
precios_muestra.append(99.0)

# Insertar en el index 1 
precios_muestra.insert(1, 60.0)

# Eliminar primera aparición 
precios_muestra.remove(49.9)

# contar veces 
print(f"Veces que hay '49.9': {precios_muestra.count(49.9)}")

# ordenar de mayor a menor 
precios_muestra.sort(reverse=True)

print(precios_muestra)
````

### 12. Consolidación de Dataset con Diccionarios

Copia el siguiente diccionario inicial:

```python
producto = {
    "sku": "LP-102",
    "categoria": "Laptops",
    "stock": 18,
    "precio": 3200.0
}
```

**Tareas:**

Realiza las siguientes operaciones utilizando métodos de diccionarios:

1. Usa `.get()` para consultar la clave `"descuento"`, devolviendo `0` por defecto.
2. Actualiza el `"stock"` a `15`.
3. Añade `"marca": "Lenovo"` mediante `.update()`.
4. Elimina `"categoria"` con `.pop()`.
5. Finalmente, recorre `.items()` con un `for` para imprimir cada clave y valor restante.

````python
producto = {
    "sku": "LP-102",
    "categoria": "Laptops",
    "stock": 18,
    "precio": 3200.0
}

# get con default 0
print(f"Buscando 'descuento' {producto.get('descuento', 0)}") # 0 valor por defecto si no encuentra la clave"


# update stock
producto.update({"stock":15})
producto

# add 'marca' with update 
producto.update({'marca': "Lenovo"})
producto

# delete 'categoria' => .pop()
producto.pop("categoria")
producto

print("\nRecorriendo el diccionario 'Producto")
for key, value in producto.items(): 
    print(f"{key} : {value}\n")
````
