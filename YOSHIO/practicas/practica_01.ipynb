```python
# Enunciado 1: Crea la función limpiar_texto(texto, estilo="titulo") que valide
# que 'texto' sea str y no esté vacío (si no, retorna "Desconocido"), quite
# espacios externos con .strip() y aplique .lower(), .upper() o .title().

def limpiar_texto(texto, estilo="titulo"): 
    if(not(isinstance(texto,str)) or  len(texto.strip()) == 0): return "Desconocido"

    texto = texto.strip()

    if(estilo.lower() in ['mayus', 'mayuscula', 'upper']): texto = texto.upper()
    elif(estilo.lower() in ['minus', 'minuscula', 'lower']): texto = texto.lower() 
    else: texto = texto.title()

    return texto
```

```python
# Enunciado 2: Crea la función convertir_numero(valor, es_entero=False, permitir_cero=True)
# que intente convertir 'valor' a int o float usando try / except (ValueError, TypeError).
# Si el número es negativo o si es cero cuando permitir_cero=False, lanza ValueError con raise.

def convertir_numero(valor, es_entero=False, permitir_cero=True): 
    try:
        valor = int(valor) if es_entero else float(valor)

        if(not(permitir_cero) and valor <= 0 ): raise ValueError("Para permitir '0' debes de especificar el argumento 'permitir_cero = True'")

        return valor 
    
    except (ValueError, TypeError) as e: 
        print(f"Error de valor '{valor}': {e}")
```

```python
# Enunciado 3: Recorre 'registros_crudos' con un bucle for. Limpia 'canal' y 'ciudad',
# lee 'descuento' de forma segura (0.0 si falta), calcula 'neto = bruto - descuento'
# y 'ingreso_por_sesion' (controlando ZeroDivisionError). Guarda los válidos en
# 'registros_limpios' y los fallidos en 'errores_log'.
registros_limpios = []
errores = []

for row in registros_crudos: 
    error = None 
    try: 
        # print(row)
        id = row['id']
        canal = row.get('canal', 'no registrado').strip().capitalize()
        ciudad = row.get('ciudad', 'no registrado').strip().capitalize() 
        bruto = float(row.get('bruto', 'no registrado'))
        descuento = float(row.get('descuento', 0.0)) 
        if(descuento < 0): raise ValueError('Descuento negativo')
        sesiones = int(row.get('sesiones', 0))
        neto = bruto - descuento
        ingreso_por_sesion = neto / sesiones

        new_row = {
            'id':id, 
            'canal':canal, 
            'ciudad':ciudad,
            'bruto': bruto,
            'descuento': descuento,
            'neto':neto, 
            'sesiones':sesiones,
            'ingreso_por_sesion': ingreso_por_sesion
        }

        registros_limpios.append(new_row)
        print(new_row)
        
    except (ValueError, TypeError, KeyError) as e:
        error = f"Error en el registro con ID {row.get("id")} | O elemento {registros_crudos.index(row)} | {e}"
    except ZeroDivisionError as e: 
        error = f"Error en el registro con ID {row.get("id")} | O elemento {registros_crudos.index(row)} | Por dividir entro 0"

    if(error != None): errores.append(error)
```
